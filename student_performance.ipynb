<c>CLEAN DATASET

COLUMN NAMES

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("\nClean column names:")
print(df.columns.tolist())

#df.head()
```

CONVERT NUMERIC COLUMNS

```python
numeric_columns = [
    "study_time_hours",
    "attendance_percent",
    "previous_grade",
    "sleep_hours",
    "final_exam_score"
]

for column in numeric_columns:
    if column in df.columns:
        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )
```

```python
import matplotlib.pyplot as plt 
import seaborn as sns
```

CHARTS

```python
plt.figure(figsize=(10,6))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Matrix")
plt.show()
```

Final mark vs no of students

```python
grade_count = df["final_grade"].value_counts()

plt.figure(figsize=(7, 4))

sns.barplot(
    x=grade_count.index,
    y=grade_count.values
)

plt.title("Final Grade Distribution")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")

plt.show()

plt.show()
```

Study Time vs Final Exam Score

```python
plt.scatter(df["study_time_hours"], df["final_exam_score"])
plt.title("Study Time vs Final Exam Score")
plt.xlabel("Study Time")
plt.ylabel("Final Exam Score")
plt.savefig("study_time_vs_final_exam_score.png")
plt.show()
```

Attendance_percent by student

```python
plt.figure(figsize=(8, 6))
plt.hist(df["attendance_percent"], bins=10, color='skyblue', edgecolor='black')
plt.title("Attendance Percentage Distribution") 
plt.xlabel("Attendance Percentage")
plt.ylabel("Number of Students")
plt.savefig("attendance_percentage_distribution.png")
plt.show()
```

study time vs final grade

```python
plt.figure(figsize=(8, 6))
sns.boxplot(data=df,x="final_grade" ,y="sleep_hours")
plt.title("Sleep Hours Distribution by Final Grade")
plt.xlabel("Final Grade")
plt.ylabel("Sleep Hours")
plt.savefig("sleep_hours_distribution.png")
plt.show()
```

Average Score — Internet Access

```python
avg_score = df.groupby("internet_access")["final_exam_score"].mean()
avg_score1 = df.groupby("extracurricular_activities")["final_exam_score"].mean()
avg_score2 = df.groupby("part_time_job")["final_exam_score"].mean()
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

axes[0].bar(avg_score.index, avg_score.values)
axes[0].set_title("Average Final Score by Internet Access")
axes[0].set_xlabel("Internet Access")
axes[0].set_ylabel("Average Final Exam Score")


axes[1].bar(avg_score1.index, avg_score1.values)
axes[1].set_title("Average Final Score by Extra Activities")
axes[1].set_xlabel("Extra Activities")
axes[1].set_ylabel("Average Final Exam Score")

axes[2].bar(avg_score2.index, avg_score2.values)
axes[2].set_title("Average Final Score by Part-Time Job")
axes[2].set_xlabel("Part-Time Job")
axes[2].set_ylabel("Average Final Exam Score")

plt.tight_layout()
plt.show()
```

SUMMARY

```python
summary = df.groupby("final_grade").agg(
    average_score=("final_exam_score", "mean"),
    average_study_time=("study_time_hours", "mean"),
    average_attendance=("attendance_percent", "mean"),
    average_sleep=("sleep_hours", "mean"),
    average_previous_grade=("previous_grade", "mean")
)

summary
```
